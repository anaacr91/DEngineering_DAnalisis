### Planteamiento del problema:

* Diseñar dos algoritmos: uno que ordene los productos, de menor a mayor precio

* buscar un producto específico para ver si está en el inventario y cuál es su precio.

###Algoritmo Ordenación Elegido: Ordenamiento por Selección.

Justificación:

* Trabaja in-place: no necesita estructuras adicionales para ordenar, solo intercambia elementos dentro de la misma lista.

* Permite ordenar fácilmente por una clave concreta, como precio

### Complejidad temporal: O(n²) polinómico cuadrático

* Se recorren todos los elementos para encontrar el mínimo, y esto se repite para cada posición de la lista.

### Complejidad Espacial O(1) constante

* Solo utiliza unas pocas variables adicionales (índices y variable para la posición del mínimo).

* No crea estructuras auxiliares de tamaño proporcional a n.

###Algoritmo Búsqueda Elegida: Búsqueda Lineal

Justificación:

* Recorre la lista desde el principio hasta el final

* Funciona aunque la lista no esté ordenada

### Complejidad temporal: O(n) lineal

* Recorre secuencialmente todos los elementos de la estructura de datos

### Complejidad espacial: O(1) constante

* Solo utiliza una variable para ir recorriendo la lista

## Implementación de los Algoritmos

```python
productos = [
    {'nombre': 'camisa', 'precio': 10},
    {'nombre': 'pantalon', 'precio': 15},
    {'nombre': 'zapatos', 'precio': 45},
    {'nombre': 'chaqueta', 'precio': 30},
    {'nombre': 'bolso','precio': 50}
]

def ordenar_productos_por_precio(productos):
    n = len(productos)
    for i in range(n - 1):
        indice_minimo = i
        for j in range(i + 1, n):
            if productos[j]['precio'] < productos[indice_minimo]['precio']:
                indice_minimo = j

        productos[i], productos[indice_minimo] = productos[indice_minimo], productos[i]
    return productos

def buscar_producto_por_nombre(productos, nombre):
      for producto in productos:
        if producto['nombre'].lower() == nombre.lower():
            return producto

productos_ordenados = ordenar_productos_por_precio(productos)

print("Productos ordenados por precio:")
for p in productos_ordenados:
    print(f"Nombre: {p['nombre']}, Precio: {p['precio']}")

print("\nBúsqueda de producto:")

nombre = "tacones"
resultado = buscar_producto_por_nombre(productos_ordenados, nombre)

if resultado is not None:
    print(f"El producto '{nombre}' está en el inventario con precio {resultado['precio']}.")
else:
    print(f"El producto '{nombre}' NO está en el inventario.")
```
