# Test de Pandas
Este notebook prueba que pandas funciona correctamente

```python
import pandas as pd
import numpy as np
import sys

print(f"Python version: {sys.version}")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

```python
# Crear un DataFrame de prueba
data = {
    'Nombre': ['Ana', 'Carlos', 'María', 'Juan', 'Laura'],
    'Edad': [25, 30, 28, 35, 27],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Bilbao'],
    'Salario': [30000, 45000, 35000, 50000, 38000]
}

df = pd.DataFrame(data)
print("DataFrame creado:")
df
```

```python
# Operaciones básicas con pandas
print("Estadísticas descriptivas:")
print(df.describe())

print("\nPromedio de edad:", df['Edad'].mean())
print("Salario máximo:", df['Salario'].max())
print("\nPersonas mayores de 28 años:")
df[df['Edad'] > 28]
```
